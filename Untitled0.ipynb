## Función 1: print()

### ¿Qué es?
La función print() permite mostrar información en pantalla.

### Sintaxis
print(valor)

### Ejemplo práctico
Podemos utilizarla para mostrar el nombre de una persona, una operación matemática o un mensaje.

### Resultado esperado
La información que escribimos dentro de print() aparece en la salida de la celda.


```python

# Ejemplo 1: mostrar un mensaje
print("Estoy aprendiendo Python")

# Ejemplo 2: mostrar un nombre
print("Mi nombre es Karen")

# Ejemplo 3: mostrar una operación matemática
print(10 + 5)
```


## 2. Función input()

### ¿Qué es?
La función `input()` es una función incorporada de Python que permite recibir información escrita por el usuario durante la ejecución de un programa. Los datos ingresados se reciben inicialmente como cadenas de texto (str).

### Sintaxis
`input(prompt)`

### Parámetros
- `prompt`: mensaje opcional que se muestra para indicar qué información debe ingresar el usuario.

### ¿Para qué se utiliza?
- Solicitar nombres y apellidos.
- Recibir edades y otros datos.
- Obtener información necesaria para realizar operaciones.
- Crear programas interactivos.

### Ejemplo de aplicación
Se puede utilizar para solicitar el nombre y la edad de una persona y posteriormente mostrar la información ingresada.

### Resultado esperado
El programa solicitará los datos al usuario y posteriormente mostrará la información proporcionada.


```python

# Ejemplo 1: Solicitar el nombre
nombre = input("Ingrese su nombre: ")
print("Hola,", nombre)

# Ejemplo 2: Solicitar la edad
edad = int(input("Ingrese su edad: "))
print("Su edad es:", edad)
```


## 3. Función type()

### ¿Qué es?
La función `type()` permite identificar el tipo de dato al que pertenece un objeto o valor en Python. Es útil para conocer si una variable almacena un número entero, un número decimal, una cadena de texto u otro tipo de dato.

### Sintaxis
`type(object)`

### Parámetros
- `object`: elemento o variable cuyo tipo de dato se desea identificar.

### ¿Para qué se utiliza?
- Identificar tipos de datos.
- Comprobar el contenido de las variables.
- Comprender cómo Python interpreta los valores.
- Facilitar la revisión y depuración de programas.

### Ejemplo de aplicación
Se puede utilizar para identificar el tipo de dato de diferentes variables que almacenan números, textos y valores booleanos.

### Resultado esperado
Python mostrará el tipo de dato correspondiente a cada variable.


```python

# Declaración de variables
nombre = "Karen"
edad = 22
estatura = 1.60
estudiante = True

# Identificación de los tipos de datos
print(type(nombre))
print(type(edad))
print(type(estatura))
print(type(estudiante))
```


## 4. Función int()

### ¿Qué es?
La función `int()` permite convertir un valor compatible en un número entero. Se utiliza principalmente para transformar cadenas de texto que contienen números enteros o convertir números decimales eliminando su parte fraccionaria.

### Sintaxis
`int(x, base=10)`

### Parámetros
- `x`: valor que se desea convertir a entero.
- `base`: base numérica utilizada para interpretar una cadena de texto. Es opcional y su valor predeterminado es 10.

### ¿Para qué se utiliza?
- Convertir datos de texto a números enteros.
- Trabajar con edades, cantidades y unidades.
- Realizar operaciones matemáticas con datos ingresados por el usuario.
- Eliminar la parte decimal de un número mediante conversión.

### Ejemplo de aplicación
Se puede utilizar para convertir una cantidad escrita como texto en un número entero y posteriormente realizar una operación matemática.

### Resultado esperado
Python mostrará el número convertido y el resultado de la operación realizada.


```python

# Ejemplo 1: Convertir texto a entero
numero = int("25")
print(numero)

# Ejemplo 2: Realizar una operación
resultado = numero + 10
print(resultado)

# Ejemplo 3: Convertir un decimal a entero
decimal = int(8.9)
print(decimal)
```


## 5. Función float()

### ¿Qué es?
La función `float()` permite convertir un valor compatible en un número de punto flotante, es decir, un número que puede representar valores decimales.

### Sintaxis
`float(x)`

### Parámetros
- `x`: valor numérico o cadena de texto que se desea convertir.

### ¿Para qué se utiliza?
- Convertir cadenas de texto en números decimales.
- Realizar operaciones con cantidades que contienen decimales.
- Representar medidas, pesos, alturas y distancias.
- Procesar datos numéricos ingresados por el usuario.

### Ejemplo de aplicación
Se puede utilizar para convertir una medida escrita como texto en un número decimal y realizar operaciones matemáticas.

### Resultado esperado
Python mostrará el valor convertido y el resultado de la operación.


```python

# Ejemplo 1: Convertir texto a decimal
precio = float("15000.50")
print(precio)

# Ejemplo 2: Sumar números decimales
resultado = precio + 2500.75
print(resultado)

# Ejemplo 3: Convertir un entero a decimal
numero = float(20)
print(numero)
```


## 6. Función str()

### ¿Qué es?
La función `str()` permite convertir diferentes tipos de datos en cadenas de texto. Facilita la unión de información y la presentación de resultados dentro de un programa.

### Sintaxis
`str(object)`

### Parámetros
- `object`: valor u objeto que se desea representar como texto.

### ¿Para qué se utiliza?
- Convertir números en cadenas de texto.
- Combinar textos con valores numéricos.
- Presentar información de manera organizada.
- Construir mensajes personalizados.

### Ejemplo de aplicación
Se puede utilizar para convertir la edad de una persona en texto y combinarla con un mensaje informativo.

### Resultado esperado
Python mostrará la información como una cadena de texto.


```python

# Ejemplo 1: Convertir un número a texto
edad = 22
edad_texto = str(edad)
print(edad_texto)

# Ejemplo 2: Combinar texto y número convertido
mensaje = "La edad del estudiante es " + str(edad) + " años"
print(mensaje)

# Ejemplo 3: Identificar el tipo de dato
print(type(edad_texto))
```


## 7. Función bool()

### ¿Qué es?
La función `bool()` permite convertir un valor en un dato booleano. Los valores booleanos solamente pueden ser True (verdadero) o False (falso).

### Sintaxis
`bool(x)`

### Parámetros
- `x`: valor que se desea evaluar y convertir en verdadero o falso.

### ¿Para qué se utiliza?
- Comprobar si un valor tiene contenido.
- Evaluar condiciones lógicas.
- Facilitar la toma de decisiones dentro de los programas.
- Trabajar con estructuras condicionales.

### Ejemplo de aplicación
Se puede utilizar para comprobar si una persona ingresó un nombre o dejó el espacio vacío.

### Resultado esperado
Python mostrará True cuando el valor sea considerado verdadero y False cuando sea considerado falso.


```python

# Ejemplo 1: Convertir un número
print(bool(1))

# Ejemplo 2: Evaluar un valor cero
print(bool(0))

# Ejemplo 3: Evaluar un texto
print(bool("Python"))

# Ejemplo 4: Evaluar un texto vacío
print(bool(""))
```


## 8. Función len()

### ¿Qué es?
La función `len()` permite conocer la cantidad de elementos que contiene un objeto. Puede utilizarse con cadenas de texto, listas, tuplas, conjuntos y diccionarios.

### Sintaxis
`len(object)`

### Parámetros
- `object`: objeto cuya cantidad de elementos se desea conocer.

### ¿Para qué se utiliza?
- Contar caracteres de una palabra o frase.
- Conocer la cantidad de elementos de una lista.
- Identificar cuántos datos contiene una colección.
- Comprobar si una colección tiene elementos.

### Ejemplo de aplicación
Se puede utilizar para contar los caracteres de un nombre y determinar cuántos productos contiene una lista.

### Resultado esperado
Python devolverá un número entero que representa la cantidad de elementos.


```python

# Ejemplo 1: Contar caracteres
nombre = "Python"
print(len(nombre))

# Ejemplo 2: Contar elementos de una lista
frutas = ["Manzana", "Pera", "Uva", "Mango"]
print(len(frutas))

# Ejemplo 3: Contar caracteres de una frase
frase = "Aprender Python es interesante"
print(len(frase))
```


## 9. Función abs()

### ¿Qué es?
La función `abs()` permite obtener el valor absoluto de un número. Esto significa que devuelve su valor sin considerar el signo negativo.

### Sintaxis
`abs(x)`

### Parámetros
- `x`: número entero, decimal o valor numérico compatible cuyo valor absoluto se desea calcular.

### ¿Para qué se utiliza?
- Calcular diferencias positivas entre cantidades.
- Determinar distancias numéricas.
- Trabajar con valores negativos y positivos.
- Realizar operaciones matemáticas.

### Ejemplo de aplicación
Se puede utilizar para calcular la diferencia absoluta entre dos temperaturas registradas.

### Resultado esperado
Python mostrará el valor absoluto del número, sin signo negativo.


```python

# Ejemplo 1: Valor absoluto de un entero negativo
print(abs(-15))

# Ejemplo 2: Valor absoluto de un decimal
print(abs(-8.5))

# Ejemplo 3: Diferencia entre dos temperaturas
temperatura_1 = 18
temperatura_2 = 25

diferencia = abs(temperatura_1 - temperatura_2)
print("Diferencia de temperatura:", diferencia)
```


## 10. Función round()

### ¿Qué es?
La función `round()` permite redondear un número al valor entero más cercano o a una cantidad específica de posiciones decimales.

### Sintaxis
`round(number, ndigits)`

### Parámetros
- `number`: número que se desea redondear.
- `ndigits`: cantidad de posiciones decimales que se desean conservar. Es opcional.

### ¿Para qué se utiliza?
- Redondear resultados de operaciones matemáticas.
- Presentar cantidades con una cantidad determinada de decimales.
- Facilitar la lectura de resultados numéricos.
- Aproximar valores en cálculos.

### Ejemplo de aplicación
Se puede utilizar para redondear el resultado de una medición o el cálculo de un área a dos posiciones decimales.

### Resultado esperado
Python mostrará el número redondeado de acuerdo con la precisión indicada.


```python

# Ejemplo 1: Redondear a un número entero
print(round(8.7))

# Ejemplo 2: Redondear a dos decimales
area = 15.6789
print(round(area, 2))

# Ejemplo 3: Redondear a tres decimales
resultado = 3.141592
print(round(resultado, 3))
```
