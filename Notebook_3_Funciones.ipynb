```python
ACTIVIDAD 3

Integrante: Karoll Nayibeth Higuera Chavez
Objetivo: Proporcionar herramientas que permitan trabajar de manera eficiente con diferentes tipos de datos en Python, facilitando su organización, combinación, transformación, selección y verificación

21. enumerate()

Descripción: Permite recorrer una colección obteniendo simultáneamente el índice y el elemento.

Objetivo: Identificar la posición y el contenido de cada elemento dentro de una colección.

Ejemplo:

estudiantes = ["Laura", "Carlos", "María", "Andrés"]

for numero, nombre in enumerate(estudiantes, start=1):
    print(numero, nombre)

Resultado esperado:

1 Laura
2 Carlos
3 María
4 Andrés

22. zip()

Descripción: Combina elementos de dos o más iterables posición por posición. El recorrido termina cuando se acaba el iterable más corto.

Objetivo: Relacionar información de diferentes listas mediante la posición de sus elementos.

Ejemplo:

estudiantes = ["Laura", "Carlos", "María"]
notas = [4.5, 3.8, 4.2]

for estudiante, nota in zip(estudiantes, notas):
    print(estudiante, nota)

Resultado esperado:

Laura 4.5
Carlos 3.8
María 4.2

23. map()

Descripción: Aplica una función a cada elemento de un iterable y devuelve un iterador con los resultados.

Objetivo: Transformar o modificar los elementos de una colección aplicando una misma operación.

Ejemplo:

precios = [10000, 20000, 30000]

nuevos_precios = list(map(lambda precio: precio * 1.10, precios))

print(nuevos_precios)

Resultado esperado:

[11000.0, 22000.0, 33000.0]

24. filter()

Descripción: Selecciona los elementos de un iterable para los que una función devuelve True.

Objetivo: Filtrar información y conservar únicamente los elementos que cumplen una condición determinada.

Ejemplo:

notas = [2.5, 4.0, 3.5, 2.8, 4.5]

aprobados = list(filter(lambda nota: nota >= 3.0, notas))

print(aprobados)

Resultado esperado:

[4.0, 3.5, 4.5]

25. any()

Descripción: Devuelve True si al menos un elemento de un iterable es verdadero.

Objetivo: Comprobar si existe al menos un elemento que cumpla una condición específica.

Ejemplo:

notas = [2.0, 2.5, 4.0, 2.8]

resultado = any(nota >= 3.0 for nota in notas)

print(resultado)

Resultado esperado:

True

26. all()

Descripción: Devuelve True si todos los elementos de un iterable son verdaderos.

Objetivo: Verificar que todos los elementos de una colección cumplan una condición determinada.

Ejemplo:

notas = [4.0, 3.5, 4.2, 3.8]

resultado = all(nota >= 3.0 for nota in notas)

print(resultado)

Resultado esperado:

True

27. open()

Descripción: Abre un archivo para leerlo, escribirlo o agregar contenido.

Objetivo: Permitir la creación, lectura y escritura de información en archivos externos.

Ejemplo:

# Crear y escribir en el archivo
with open("estudiante.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Nombre: Laura\n")
    archivo.write("Edad: 20 años")

# Abrir y leer el archivo
with open("estudiante.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)

Resultado esperado:

Nombre: Laura
Edad: 20 años

28. help()

Descripción: Muestra documentación y ayuda sobre una función, clase o módulo.

Objetivo: Consultar información sobre el funcionamiento, las características y la utilización de las herramientas de Python.

Ejemplo:

help(len)

Resultado esperado:

Muestra la documentación de la función len(), incluyendo su descripción y la manera de utilizarla.

29. dir()

Descripción: Devuelve una lista de nombres de atributos y métodos disponibles en un objeto o módulo.

Objetivo: Identificar las propiedades y operaciones que se pueden utilizar con un objeto.

Ejemplo:

palabra = "Python"

print(dir(palabra))

Resultado esperado:

Muestra una lista de métodos y atributos disponibles para las cadenas de texto, entre ellos:

capitalize
lower
upper
replace
strip

30. isinstance()

Descripción: Comprueba si un objeto pertenece a un tipo o clase determinada. Devuelve True o False.

Objetivo: Verificar el tipo de dato de una variable para determinar si corresponde a una clase específica.

Ejemplo:

edad = 20

nombre = "Laura"

print(isinstance(edad, int))
print(isinstance(nombre, str))
print(isinstance(edad, str))

Resultado esperado:

True
True
False
```
