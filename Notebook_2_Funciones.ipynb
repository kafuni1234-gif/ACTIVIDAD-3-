# **Función 11: max()**
**Descripción**

La función **max()** permite obtener el valor más alto dentro de una colección de datos o entre varios valores suministrados.

**Sintaxis**

max(iterable)

**Ejemplo**

```python
calificaciones = [3.5, 4.2, 4.8, 3.9, 4.5]

mayor = max(calificaciones)

print("La calificación más alta es:", mayor)
```

# **Función 12: min()**
**Descripción**

La función **min()** devuelve el valor más pequeño de una secuencia de datos.

**Sintaxis**

min(iterable)

**Ejemplo**

```python
calificaciones = [3.5, 4.2, 4.8, 3.9, 4.5]

menor = min(calificaciones)

print("La calificación más baja es:", menor)
```

# **Función 13: sum()**
**Descripción**

La función **sum()** se utiliza para sumar todos los elementos numéricos de una lista o colección.

**Sintaxis**

sum(iterable)

**Ejemplo**

```python
ventas = [250, 320, 180, 450, 300]

total_ventas = sum(ventas)

print("Total de ventas:", total_ventas)
```

# **Función 14: pow()**
**Descripción**

La función **pow()** calcula la potencia de un número elevando una base a un exponente.

**Sintaxis**

pow(base, exponente)

**Ejemplo**

```python
resultado = pow(5, 3)

print("5 elevado a la 3 es:", resultado)
```

# **Función 15: divmod()**
**Descripción**

La función **divmod()** devuelve simultáneamente el cociente y el residuo de una división.

**Sintaxis**

divmod(a, b)

**Ejemplo**

```python
cociente, residuo = divmod(27, 4)

print("Cociente:", cociente)
print("Residuo:", residuo)
```

# **Función 16: sorted()**
**Descripción**

La función **sorted()** ordena los elementos de una secuencia en orden ascendente por defecto.

**Sintaxis**

sorted(iterable)

**Ejemplo**

```python
numeros = [8, 3, 10, 1, 5]

ordenados = sorted(numeros)

print("Lista ordenada:", ordenados)
```

# **Función 17: range()**
**Descripción**

La función **range()** genera una secuencia de números enteros dentro de un intervalo determinado.

**Sintaxis**

range(inicio, fin)

**Ejemplo**

```python
for numero in range(1, 6):
    print(numero)
```

# **Función 18: list()**
**Descripción**

La función **list()** permite crear listas o convertir otros tipos de datos en listas.

**Sintaxis**

list(iterable)

**Ejemplo**

```python
tupla = (10, 20, 30, 40)

lista = list(tupla)

print(lista)
print(type(lista))
```

# **Función 19: tuple()**
**Descripción**

La función **tuple()** convierte una secuencia en una tupla, estructura de datos ordenada e inmutable.

**Sintaxis**

tuple(iterable)

**Ejemplo**

```python
lista = [10, 20, 30, 40]

tupla = tuple(lista)

print(tupla)
print(type(tupla))
```

# **Función 20: set()**
**Descripción**

La función **set()** crea conjuntos, estructuras que almacenan elementos únicos y eliminan automáticamente los duplicados.

**Sintaxis**

set(iterable)

**Ejemplo**

```python
datos = [1, 2, 2, 3, 3, 4, 4, 5]

conjunto = set(datos)

print("Conjunto generado:", conjunto)
```
